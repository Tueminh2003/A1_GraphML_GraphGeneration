## 6. Show the geometry

```python
Topology.Show(objects,
              faceColor=[210,210,250],
              faceOpacity=1,
              edgeColor="white",
              edgeWidth=3,
              showVertices=False,
              backgroundColor="white",
              width=800,
              height=600,
              renderer = renderer)
```

```python
# Create a graph where each vertex represents a room (center of each topology)
# First, let's create a cluster from all the objects
cluster = Cluster.ByTopologies(objects)

# Create a graph from the cluster where vertices represent room centers
# and edges connect adjacent rooms
g2 = Graph.ByTopology(topology=cluster, 
                      direct=False, 
                      directApertures=False, 
                      viaSharedTopologies=True, 
                      viaSharedApertures=False, 
                      toExteriorTopologies=False, 
                      toExteriorApertures=False, 
                      toContents=False, 
                      toOutposts=False, 
                      useInternalVertex=True, 
                      tolerance=0.0001)

print("Graph created successfully!")
print(f"Number of vertices (rooms): {len(Graph.Vertices(g2))}")
print(f"Number of edges (connections): {len(Graph.Edges(g2))}")
```

## 8. Assign Visual Attributes to Graph Vertices and Edges

```python
# Assign visual attributes to vertices (rooms)
vertices = Graph.Vertices(g2)
for v in vertices:
    d = Dictionary.ByKeysValues(["size", "color"], [18, "red"])
    v = Topology.SetDictionary(v, d)

# Assign visual attributes to edges (connections between rooms)
edges = Graph.Edges(g2)
for e in edges:
    d = Dictionary.ByKeysValues(["width", "color"], [4, "black"])
    e = Topology.SetDictionary(e, d)

print(f"Visual attributes assigned to {len(vertices)} vertices and {len(edges)} edges")
```

```python
# Visualize the graph
Graph.Show(g2,
           vertexSize=18,
           vertexColor="red",
           edgeWidth=4,
           edgeColor="black",
           backgroundColor="white",
           width=800,
           height=600,
           renderer=renderer)
```
